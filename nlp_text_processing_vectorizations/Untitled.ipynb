```python
class car:
    def __init__(self,r,s,eng:str):
        self.r = r
        self.s = s
        self.eng = eng
    def disp(self):
        print(self.r,self.eng,self.s)

class bmw(car):
    def __init__(self, r,s,eng, yr, cap:str):
        self.yr = yr
        self.cap = cap
        super().__init__(r,s,eng)
    def disp(self):
        print(self.r,self.eng,self.s,self.cap,self.yr)
```

```python
cr = car(1000,5,"hyn")
cr.disp()
```

```python
cr_bmw = bmw(1500,2,"bmw", 2005, "sport")
cr_bmw.disp()
```
