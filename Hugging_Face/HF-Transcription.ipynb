#### What is Audio Transcription?

Transcription means converting spoken audio into written text. Example: You have a .wav or .mp3 file of someone speaking → the model listens and outputs the words as text. This is essential for speech‑to‑text applications like subtitles, meeting notes, or voice assistants.

#### ASR Automatic Speech Recognition:

It’s the technology that converts spoken language (audio) into written text. Think of it as the “ears” of an AI system — it listens to speech and transcribes it. Modern ASR systems use deep learning models (like Whisper, Wav2Vec2, etc.) trained on huge amounts of audio + text pairs.

- **Audio Input** → You provide a .wav or .mp3 file.
- **Feature Extraction** → The model converts sound waves into spectrograms (visual representation of frequencies).
- **Neural Network Processing** → Transformer‑based ASR models map audio features to text tokens.
- **Decoding** → Tokens are converted back into readable words.
- **Output** → You get the transcription (text) plus optional timestamps.
