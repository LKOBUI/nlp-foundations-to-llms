## Query Rewriting in RAG

Query rewriting is the process of transforming a user’s original query into a more effective version so that the retrieval system can fetch the most relevant chunks before passing them to the LLM. It’s like teaching the system to “understand what you really meant” and match it better against stored knowledge.

User queries are often short, ambiguous, or incomplete. Retrieval models rely on embeddings, which may miss intent if the query is vague. Rewriting improves recall and precision, ensuring the right documents are retrieved.

Query Expansion is nothing Adds extra terms or variations to the original query without replacing it.

## Re-ranking in RAG

Re-ranking is an advanced technique in Retrieval-Augmented Generation (RAG) where the system reorders retrieved documents to ensure the most relevant ones are placed at the top before passing them to the LLM.

#### How It Works
- Initial Retrieval:
    - A vector database (FAISS, Pinecone, Chroma, etc.) returns the top k chunks based on similarity scores. These scores are often based on cosine similarity of embeddings. These can be considered as the primary models. These models are Bi-encoders work on the cosine or dot product method.
- Re-ranking Step:
    - A secondary model (often a cross-encoder or LLM) evaluates the query–document pairs more deeply. Instead of just embedding similarity, it considers semantic alignment, context, and intent. The documents are then reordered by relevance.
- Final Selection:
    - The top n re-ranked documents are injected into the LLM prompt for generation.

#### Techniques for Re-ranking:
- Cross-Encoders (e.g., BERT, MiniLM)
- LLM-as-a-Judge: Use an LLM to score retrieved chunks based on query relevance
- Hybrid Re-ranking: 
    - Hybrid re-ranking combines multiple retrieval signals (like keyword search, embeddings, and semantic scoring) to reorder documents so the most relevant ones appear at the top. Instead of relying on a single method, it blends strengths from different approaches to balance recall (getting all potentially relevant docs) and precision (keeping only the most relevant ones).

## GraphRAG :

GraphRAG is an advanced Retrieval-Augmented Generation (RAG) method developed by Microsoft Research that uses knowledge graphs instead of plain text chunks to improve reasoning over complex datasets. Unlike baseline RAG, which relies on vector similarity search, GraphRAG builds a structured graph of entities, relationships, and community summaries, enabling LLMs to answer global, multi-hop, and sensemaking queries more effectively.

### What Makes GraphRAG Different from Vector RAG

Retrieves text chunks based on embedding similarity. Works well for localized queries (e.g., “What’s the warranty on Model X?”). Struggles with global or multi-hop reasoning (e.g., “What are the main themes across 10 years of research papers?”).

## Visualizing with t‑SNE in RAG:

In Retrieval‑Augmented Generation (RAG), embeddings are the backbone: they represent documents and queries in vector space. But since embeddings are usually hundreds or thousands of dimensions, it’s hard to “see” how well they cluster. That’s where t‑SNE (t‑Distributed Stochastic Neighbor Embedding) comes in — it reduces high‑dimensional embeddings into 2D or 3D plots so you can visually inspect how your chunks, queries, and retrieved results relate.

#### Why Use t‑SNE in RAG:

- Cluster inspection → See if semantically similar chunks group together.
- Quality check → Spot noisy embeddings or poor chunking strategies.
- Model comparison → Compare embeddings from different encoders (e.g., OpenAI vs. BERT).
- Debug retrieval → Verify whether queries land near the right document clusters.
