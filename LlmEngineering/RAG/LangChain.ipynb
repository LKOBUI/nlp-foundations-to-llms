## Explaining Encoder / Embedding Models:

- Encoders → understand and represent meaning.
- Embedding models → output dense vectors used for similarity search.
- Together, they form the “Encoding LLM” in your RAG architecture — the part that converts text into retrievable numerical form.

#### What an Encoder / Embedding Model Does?

It takes text as input (a sentence, paragraph, or document). It outputs a `vector embedding` — a list of numbers that capture the semantic meaning of the text. These` embeddings are stored in a vector database` (like `Chroma`, `FAISS`, `Pinecone`, or `Weaviate`). When a user asks a question, the system converts the query into a vector and finds the most similar stored vectors — that’s how retrieval works.

#### Workflow Steps:

- **Read in the documents** → LangChain can load files from folders (PDFs, text, Word, etc.).
- **Divide into chunks** → Large documents are split into smaller overlapping pieces (e.g., 500 characters with 50 overlap) so retrieval is more precise.
- **Vectorize and store** → Each chunk is passed through an encoder/embedding model to produce a vector. These vectors are stored in a vector database for fast similarity search.

- `Encoders` (Embedding Models): These are the models that convert text into vectors
    - OpenAI → text-embedding-3-small, text-embedding-3-large
    - Google → gemini-embedding-001
    - Hugging Face → sentence-transformers/all-MiniLM-L6-v2
- `Vectorstores` (Databases for embeddings): Where the vectors are stored and searched
    - Open-source: Chroma, Qdrant, FAISS (in-memory, fast but not persistent).
    - Paid & scalable: Pinecone, Weaviate (cloud-hosted, enterprise scale).
    - Mainstream DBs: Postgres, MongoDB, Elastic (extended with vector search capabilities).

#### Choice of vector database is independent of the embedding model. It’s mainly a cost vs. performance tradeoff:

- FAISS → fast, local experiments.
- Pinecone/Weaviate → scalable production systems.
- Chroma/Qdrant → open-source, flexible options.

- LangChain orchestrates the pipeline (load → chunk → embed → store).
- You choose which encoder (to generate embeddings) and which vectorstore (to hold them).

### Two key abstractions from LangChain:

LangChain’s two key abstractions are the LLM (Large Language Model interface) and the Retriever. Together, they form the backbone of Retrieval-Augmented Generation (RAG), enabling models to combine reasoning with external knowledge.

* LLM Interface:
    - LangChain Provides a standardized way to interact with generative models like GPT, LLaMA, Claude, or Hugging Face models. LangChain Capabilities Accepts prompts and returns generated text.Abstracts away API differences between providers.
    - Instead of writing custom code for each model, LangChain lets you swap LLMs easily without changing your pipeline logic. Example: You can switch from OpenAI GPT‑4 to Anthropic Claude by changing one line of configuration.
* Retriver:
    - Fetches relevant documents or data chunks from a knowledge base (vector store, SQL DB, PDFs, etc.) given a user query.
    - Similarity search (find closest embeddings). Advanced strategies like MMR (Maximum Marginal Relevance) to balance relevance and diversity. Metadata filtering (e.g., restrict results to a specific source or date). Score thresholds to ensure only high-quality matches are returned.

#### How They Work Together?

- LLM alone: 
    - Generates text but limited to its training data and context size. 
- Retriever alone
    -  Finds relevant documents but cannot synthesize or explain them. 
- LLM + Retriever (RAG):
    - User asks a question.
    - Retriever finds relevant documents.
    - LangChain assembles a prompt combining the query + retrieved context.
    - LLM generates a grounded, context-aware answer.
