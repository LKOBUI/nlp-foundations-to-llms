#### Prompt Engineering
##### Two Types of Prompts in LLMs

- System prompt ensures consistency (tone, persona, constraints).
    - Sets the rules of engagement.
    - Defines the model’s role, task, and tone.
- User prompt drives the specific task.
    - The actual instruction or question from the user.

```python
# Example: System vs User Prompts with OpenAI
from openai import OpenAI
from dotenv import load_dotenv
import os

# Load your API key from .env
load_dotenv()
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

import requests
from bs4 import BeautifulSoup
#from scraper import fetch_website_contents

from IPython.display import Markdown, display
```

```python
# Define system + user prompts
messages = [
    {"role": "system", "content": "You are a witty assistant who explains concepts in a fun, easy-to-understand way."},
    {"role": "user", "content": "Explain cosine similarity with a simple example."}
]

# Make the API call
response = client.chat.completions.create(
    model="gpt-4.1-mini",   # or another available model
    messages=messages
)

# Print the model's reply
print(response.choices[0].message.content)
```

```python
def fetch_website_contents(url):
    response = requests.get(url)
    soup = BeautifulSoup(response.text, "html.parser")
    return soup.get_text()
```

```python
ed = fetch_website_contents("https://www.electrical4u.com/differential-protection-of-transformer-differential-relays/#google_vignette")
```

```python
system_prompt = """
You are a snarky assistant that analyzes the contents of a website given knowledge some electrical field,
and able to provide sort summary with solid explanations & mathematically. if possible create diagram also. 
Your responsibility first check website contains electrical and mathematical knowledge. If given URL
not from electricals or mathematical field jut warn them I am electrical engineer only and my knowledge is limited."""

user_prompt_prefix = """
what is the meaning of differential protections and how it is helpful for transformer.
"""
```

```python
def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + website}
    ]
```

```python
response = client.chat.completions.create(
        model = "gpt-4.1-mini",
        messages = messages_for(ed)
    )
```

```python
#print(response.choices[0].message.content)
display(Markdown(response.choices[0].message.content))
```

```python
def summarize(url):
    website = fetch_website_contents(url)
    response = client.chat.completions.create(
        model = "gpt-4.1-mini",
        messages = messages_for(website)
    )
    return response.choices[0].message.content
```

```python
# A function to display this nicely in the output, using markdown

def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
```

#### What’s an Endpoint?

An endpoint is simply the URL where you send your request to talk to an API. Think of it as the “doorway” into the service. For OpenAI, endpoints like /chat/completions let you send prompts and receive model outputs.

#### What is the openai package?

It's known as a Python Client Library. It's nothing more than a wrapper around making this exact call to the http endpoint.

#### Why it’s called a Python Client Library ?

A client library is code that runs on your computer (the client) and makes it easier to talk to a remote service (the server).

The openai Python package is just a wrapper around HTTP requests to the OpenAI API endpoints. Instead of manually building JSON and sending it with requests.post(), you can call neat Python functions like

client.chat.completions.create(model="gpt-4.1-mini", messages=messages)

→ Under the hood, it’s still sending JSON to the server, but you don’t have to deal with the low‑level details.

#### So What’s a Server Library in this case ?

A server library is code that runs on the server side (the machine hosting the service). It helps developers build and expose APIs, handle requests, manage authentication, and return responses. Examples: Flask, FastAPI, Django (for Python) — these are server libraries/frameworks used to build web services.

⚡ Analogy
Client library: Like a phone app that makes calling easy — you press a button instead of dialing raw numbers.

Server library: Like the telecom infrastructure that actually routes and connects the call.

#### Create Endpoint Using Json

```python
headers = {"Authorization": f"Bearer {os.getenv("OPENAI_API_KEY")}", "Content-Type": "application/json"}
url = "https://eshop.se.com/in/blog/post/differential-relay-types-and-working-principle.html?srsltid=AfmBOoqNHEdEFy9iE9uPV9HCV06_o94mqCWnfPHex5J1f_b3ggTF_TYN"
payload = {
    "model": "gpt-5-nano",
    "messages": [
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt_prefix + url}
        ]
}

print(payload)


response = requests.post(
    "https://api.openai.com/v1/chat/completions",
    headers=headers,
    json=payload
)

print(response.json())
```

```python
display(Markdown(response.json()["choices"][0]["message"]["content"]))
```

#### What is POST Request actually doing ?

Sends your payload to OpenAI’s chat completions endpoint. The endpoint processes your request and generates a response from the model.

#### Tokenizing with code

- **Tokenization in NLP (general)**: Splitting text into smaller units (words, subwords, characters) so models can process them.
- **Tokenizing with code (OpenAI context)**: Using a programming interface (tiktoken) to see exactly how a model like gpt-4.1-mini represents text internally as tokens (integers).
    - Each token corresponds to a piece of text (could be a whole word, part of a word, or punctuation).
    - The model doesn’t "see" raw text — it only works with these token IDs.

#### How it works

```python
import tiktoken

# Loads the tokenizer rules specific to that model.

encoding = tiktoken.encoding_for_model("gpt-4.1-mini")
```

```python
# Converts the sentence into a list of integers (token IDs).
tokens = encoding.encode("Ram getting grass but he is not cow he is human. ")
```

```python
# Shows the mapping between each integer and its decoded text fragment.

for token_id in tokens:
    print(token_id, encoding.decode([token_id]))
```

```python
encoding.decode([4809])
```

Demonstrates how one token ID corresponds to a specific string. You’re seeing the actual implementation for LLMs, where text is mapped to integer IDs that the model consumes.

#### Why Tokenization matters models ?
**Efficiency**: Models process tokens, not raw text. Tokenization defines input length limits (e.g., 8k or 32k tokens).

**Granularity**: Unlike word-level tokenization in traditional NLP, OpenAI uses Byte Pair Encoding (BPE) or similar subword methods. That’s why "banoffee" might split into multiple tokens.

#### What is meaning of 8k or 32k tokens?

Models don’t read characters or words directly. A language model like GPT doesn’t see "Hi my name is Ed..." as letters or words. Instead, it sees a sequence of token IDs (integers). Each token is a chunk of text (could be a whole word, part of a word, or punctuation).

**Token limits = maximum context window:**

- Every model has a context window size — the maximum number of tokens it can handle in one input.
    - GPT‑4.1‑mini might allow 8,000 tokens.
    - GPT‑4‑turbo can go up to 128,000 tokens.
If your text exceeds that limit, the model cannot process it all at once.

#### The Illusion of ‘memory':

Models don’t have persistent memory by default. Every time you send a request to the API, the model only sees the tokens you include in that request. It doesn’t automatically remember past requests or your API key.

The illusion comes from your environment setup. When you store your API key in a `.env` file and load it with `dotenv`, it feels like the system “remembers” your key. But in reality: Your operating system or environment variables are holding onto it. The model itself has no awareness of your key or past inputs. So the `“memory”` is external. It’s your code, files, and environment that provide persistence — not the OpenAI model.

**Q. Why this matters?**

If you restart your notebook or clear your environment, the `model won’t magically recall your API` key or previous inputs. To make the model “remember” across turns, you have to` manually feed prior conversation tokens` back in (that’s what context windows are for). True long‑term memory (like what I have with you here) is a separate feature built outside the raw API.

**Q. Why you need to feed prior tokens back?**

Each API call is independent. The `model doesn’t remember what you said before`. To make it feel like memory, you must `include the previous conversation text` again in the next request. That’s what “manually feeding prior conversation tokens back” means.

#### Collect prior messages:

You store the conversation history in your own code (list, database, or file). Example:

```python
conversation = [
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "Hi, who won the 2018 World Cup?"},
    {"role": "assistant", "content": "France won the 2018 FIFA World Cup."},
    {"role": "user", "content": "Who scored in the final?"}
]

# Send history back with the new query

response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=conversation
)
```

Each message is tokenized. The total number of tokens across all messages must fit within the model’s context window (e.g.,` 8k, 32k, 128k tokens`). That’s why token limits matter: if your history grows too long, you must summarize or truncate old messages.

**Q. "When you call the API, you don’t just send the latest user message. You send the entire conversation list" The meaning of this line each and every time conversations list become increase (past + latest conversations). So in this situations tokenization's also increases**

**Conversation list growth**: Each time you send a new message, you append it to the conversation list along with the assistant’s reply. So yes, the list keeps growing:
- Turn 1 → [system, user, assistant]
- Turn 2 → [system, user, assistant, user, assistant]
- Turn 3 → [system, user, assistant, user, assistant, user, assistant]
- Example:
    - One short user message = ~20 tokens.
    - After 10 turns, you might have ~200–300 tokens.
    - After 100 turns, you could easily hit thousands of tokens.

#### Context window limit

Each model has a maximum number of tokens it can handle `(e.g., 8k, 32k, 128k)`. If your conversation history exceeds that, you must:
- Truncate older messages (drop them).
- Or summarize them into fewer tokens.
- Otherwise, the API will reject the request.

every time you send the whole conversation list, the tokenization increases because you’re re‑encoding past + latest messages. That’s the trade‑off: you get continuity, but you must manage token limits.

#### Example: Growing Conversation Token Count

```python
import tiktoken

# Load tokenizer for your model
encoding = tiktoken.encoding_for_model("gpt-4.1-mini")

# Start with system prompt
conversation = [
    {"role": "system", "content": "You are a helpful assistant."}
]

# Function to calculate token count
def count_tokens(conversation):
    # Concatenate all messages into one string
    text = ""
    for msg in conversation:
        text += msg["role"] + ": " + msg["content"] + "\n"
    return len(encoding.encode(text))

# Simulate conversation turns
turns = [
    {"role": "user", "content": "Hi, who won the 2018 World Cup?"},
    {"role": "assistant", "content": "France won the 2018 FIFA World Cup."},
    {"role": "user", "content": "Who scored in the final?"},
    {"role": "assistant", "content": "Mario Mandzukic scored an own goal, and Antoine Griezmann, Paul Pogba, and Kylian Mbappé scored for France."},
    {"role": "user", "content": "Thanks! Can you summarize the match in 2 sentences?"}
]

# Add turns one by one and check token count
for i, msg in enumerate(turns, start=1):
    conversation.append(msg)
    print(f"After turn {i}, token count = {count_tokens(conversation)}")
```
