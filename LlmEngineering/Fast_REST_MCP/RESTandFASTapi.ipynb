# What is REST ?

* **REST** is a term coined by Roy Fielding to describe an **architecture style** of networked systems. REST is an acronym standing for **R**epresentational **S**tate **T**ransfer.
* REST is a set of design criteria and not the physical structure (architecture) of the system.
* This style was initially described by Roy Fielding in 2000.


## REST architecture

![REST architecture](image.png)

REST (Representational State Transfer) is an architectural style for designing networked applications that treats resources (identified by URIs) as the central concept and uses a small set of well‑defined operations over HTTP to transfer **representations** of those resources (JSON, XML, text, HTML).

* **Resource:** Any information or object that can be named (a user, an order, a sensor reading). Each resource has a unique **URI**.
* **Representation:** A resource’s state transferred between client and server (for example, JSON or XML). The client receives a representation and can act on it.
* **Client-Server separation:** Clients handle user interface and user state; servers manage data and application logic. This separation improves portability of the client and scalability of the server.
* **Statelessness:** Each request from client to server must contain all information needed to understand and process the request; the server does not store client session state between requests. This simplifies server design and enables easier scaling.
* **Cacheability:** Responses must declare whether they are cacheable. Proper caching reduces client-server interactions and improves performance.

### Architectural constraints (the rules that define REST)

1. **Uniform interface** — a small, consistent set of operations and conventions (e.g., use HTTP methods semantically: `GET` to retrieve, `POST` to create, `PUT` to replace, `PATCH` to modify, `DELETE` to remove). This constraint decouples client and server and simplifies interaction.
2. **Stateless interactions** — each request is self-contained; servers don’t keep client session state.
3. **Cacheable responses** — responses must indicate cacheability to improve efficiency.
4. **Client-server** — separation of concerns between UI and data/storage.
5. **Layered system** — an architecture composed of hierarchical layers (e.g., load balancers, proxies, gateways) so components don’t need to know beyond the next layer.
6. **Code on demand (optional)** — servers can optionally send executable code (like JavaScript) to extend client functionality. This is the only optional constraint.

### How REST maps to HTTP in practice

* **Resource URIs:** `/users/123`, `/orders/2026-09-13` — identify resources.
* **HTTP methods:**
    * `GET` — read a representation (safe, idempotent).
    * `POST` — create a subordinate resource or trigger processing (not necessarily idempotent).
    * `PUT` — replace a resource (idempotent).
    * `PATCH` — partially update a resource.
    * `DELETE` — remove a resource.
    * `HEAD` — Retrieve only the headers of a resource, not the body. When you want metadata (like content type, length, last modified date) without downloading the full resource. Saves bandwidth when you only need meta‑information or want to check if a resource exists.
    * `OPTIONS` — Discover which HTTP methods are supported by a resource. When a client wants to know what operations are allowed before making a request. Useful for API exploration, debugging, or implementing CORS (Cross‑Origin Resource Sharing), where browsers send an OPTIONS preflight request to check permissions.

`HEAD & OPTIONS` less commonly used methods.

* **Status codes:** Use standard HTTP status codes to indicate outcome (`200 OK`, `201 Created`, `204 No Content`, `400 Bad Request`, `401 Unauthorized`, `404 Not Found`, `500 Internal Server Error`).
* **Content negotiation:** Clients and servers negotiate representation formats via headers (e.g., `Accept: application/json`).

### Example

* **Request:** `GET /weather/UK/London` with `Accept: application/json`
* **Response:** `200 OK` with JSON body `{ "city": "London", "temp_c": 15, "unit": "C" }`

This shows a client retrieving a resource representation (weather for London) via a URI and HTTP GET.


#### why REST is called Representational State Transfer?

![REST is called Representational State Transfer](image-1.png)

* **Representational** $\rightarrow$ The client doesn't directly manipulate the resource itself; instead, it works with a *representation* of the resource (like JSON, XML, or text). That representation contains the information needed to identify, modify, or delete the resource.
* **State** $\rightarrow$ The *application state* (what the client is doing, like browsing weather reports or editing a profile) is kept on the client side. The server only manages *resource state* (the actual data, like the weather report for London).
* **Transfer** $\rightarrow$ The client sends its state to the server through HTTP requests, and the server responds with resource representations. This back-and-forth is the "transfer" part.


## REST API restaurant analogy

![REST API restaurant analogy ](image-2.png)

* **User (Customer):** Represents the client (like your app or browser). The user wants something — for example, data or a service.
* **Waiter (API):** Acts as the middleman. The waiter takes the user’s request, translates it into something the kitchen understands, and delivers the response back. This is exactly what an API does — it takes client requests and communicates with the server/application.
* **Kitchen (Application/Server):** The place where the actual work happens. The kitchen prepares the food (data or service) based on the request. The client doesn’t interact directly with the kitchen; everything goes through the waiter (API).
* **Food (Data/Resources):** The output — coffee, fries, ice cream in the analogy — corresponds to the resources or data returned by the application (like tweets from Twitter’s REST API).

#### A REST API is like the **waiter** in a restaurant:

* It **receives requests** (GET, POST, PUT, DELETE).
* It **forwards them** to the application (kitchen).
* It **returns responses** (data, status codes) back to the client (user).



## FastAPI

***Q. Now I understand how REST APIs (like POST, GET, PUT, and DELETE) work. My model runs on the server, and I can use a REST API from the client side to communicate with it. As I understand it, a REST API acts like a postman between the client and the server. However, I don’t quite understand where FastAPI fits into this scenario.***

#### Where FastAPI fits

* **FastAPI is a Python framework** for building REST APIs quickly and efficiently.
* It provides the tools to define endpoints (`/predict`, `/train`, `/delete`) and automatically handles HTTP methods (GET, POST, PUT, DELETE).
* It takes care of parsing requests, validating inputs, returning JSON responses, and even generating interactive API docs (Swagger UI).
* In your scenario:
    * **Server side:** Your ML model runs inside a Python process.
    * **FastAPI:** Wraps your model with REST endpoints.
    * **Client side:** Apps, scripts, or Postman send HTTP requests to those endpoints.

#### Q. Where FastAPI exist? 

FastAPI runs on the server side. Start a FastAPI application (usually with `uvicorn main:app --reload`). Inside that app, you define REST endpoints (`/predict`, `/train`, `/delete`). Now ML model is loaded in memory on the server, and FastAPI acts as the *interface* to expose it.


![user interacts with the app](image-3.png)

#### Wrapping a Model with FastAPI (Example only to understand.)

#### Server (FastAPI)

```python
from fastapi import FastAPI
from pydantic import BaseModel
import joblib

# Load your trained model
model = joblib.load("wine_quality_model.pkl")

app = FastAPI()

# Define input schema
class WineFeatures(BaseModel):
    fixed_acidity: float
    volatile_acidity: float
    citric_acid: float
    residual_sugar: float
    chlorides: float
    free_sulfur_dioxide: float
    total_sulfur_dioxide: float
    density: float
    pH: float
    sulphates: float
    alcohol: float

@app.post("/predict")
def predict_quality(features: WineFeatures):
    data = [[
        features.fixed_acidity, features.volatile_acidity, features.citric_acid,
        features.residual_sugar, features.chlorides, features.free_sulfur_dioxide,
        features.total_sulfur_dioxide, features.density, features.pH,
        features.sulphates, features.alcohol
    ]]
    prediction = model.predict(data)[0]
    return {"predicted_quality": int(prediction)}
```

#### Client (Postman / Python requests)

```python
import requests
response = requests.post("http://localhost:8000/predict", json={
    "fixed_acidity": 7.4,
    "volatile_acidity": 0.7,
    "citric_acid": 0.0,
    "residual_sugar": 1.9,
    "chlorides": 0.076,
    "free_sulfur_dioxide": 11.0,
    "total_sulfur_dioxide": 34.0,
    "density": 0.9978,
    "pH": 3.51,
    "sulphates": 0.56,
    "alcohol": 9.4
})
print(response.json())
```

- Server (FastAPI): Receives the request, passes data to your model, and returns the prediction.
- REST API: The communication protocol between them.

# Swagger docs

***Q. What is Swagger docs and what use in ML deployment? how FastAPI automatically generates Swagger docs?***

#### What are Swagger docs?

* **Swagger (now part of OpenAPI)** is a standard for describing REST APIs.
* It provides an **interactive documentation page** where you can:
    * See all available endpoints (`/predict`, `/train`, `/delete`).
    * Understand what inputs each endpoint expects (query params, JSON body).
    * Try out requests directly in the browser and see responses.

Think of it as a **live Postman built into your API**.

#### Why useful in ML deployment?

When you deploy a machine learning model:

* **You want other developers, testers, or even non-technical stakeholders to know how to call your API.**
* Swagger docs make it easy to:
    * Test predictions without writing client code.
    * Share API usage with teammates.
    * Debug input/output formats quickly.
    * Reduce errors (since the docs are auto-generated from your code).

For example, if your model expects a JSON payload with features, Swagger shows the schema automatically.

#### How FastAPI generates Swagger docs

FastAPI is built on **OpenAPI**. When you define endpoints and input models (using Pydantic), FastAPI automatically:

* Creates an **OpenAPI specification**.
* Serves interactive docs at:
    * `http://localhost:8000/docs` $\rightarrow$ Swagger UI
    * `http://localhost:8000/redoc` $\rightarrow$ ReDoc (another doc style)

No extra setup needed — it's built-in.

## Example

```python
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI()

class InputData(BaseModel):
    age: int
    income: float

@app.post("/predict")
def predict(data: InputData):
    return {"prediction": "approved" if data.income > 50000 else "denied"}
```
- Run with
`uvicorn main:app --reload
`

* Open browser at `http://localhost:8000/docs` $\rightarrow$ You'll see Swagger UI with:
    * Endpoint `/predict`
    * Input schema (`age`, `income`)
    * A "Try it out" button to send test requests
