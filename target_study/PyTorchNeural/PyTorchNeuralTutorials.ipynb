### Optimizers in Neural Networks

An Optimizer is an algorithm used to update neural network parameters (weights and biases) to minimize the loss function.

In simple words:
- Neural network makes predictions.
- Loss function calculates prediction errors.
- Backpropagation calculates gradients.
- Optimizer updates weights and biases using those gradients.
- The process repeats until the model learns.

#### Neural network training workflow

1. **Forward Propagation**  
   `Input → Neural Network → Prediction`
   
   ↓

2. **Loss Function**  
   Calculate prediction error
   
   ↓

3. **Backpropagation**  
   `loss.backward()` → Calculate gradients
   
   ↓

4. **Optimizer**  
   Uses gradients to update weights
   
   ↓

5. **Repeat Training**  
   Next forward pass with updated weights


### Optimizer Mathematical understanding

Suppose the loss function is:

$$L(w) = (w - 5)^2$$

Initial weight:

$$w = 2$$

**Step 1: Loss calculation**

$$L(2) = (2 - 5)^2 = 9$$

**Step 2: Backpropagation**

Calculate the derivative:

$$\frac{dL}{dw} = 2(w - 5)$$

$$= 2(2 - 5) = -6$$

The gradient is $-6$.

**Step 3: Optimizer updates the weight**

Assume SGD with learning rate $0.1$:

$$w_{new} = w_{old} - \eta \times \text{gradient}$$

$$= 2 - 0.1(-6)$$

$$\boxed{w_{new} = 2.6}$$

The optimizer has updated the weight from $2$ to $2.6$.


````python
import torch
import torch.nn as nn

# Initial weight
w = nn.Parameter(torch.tensor(2.0))

# Optimizer
optimizer = torch.optim.SGD([w], lr=0.1)

# Loss function
for epoch in range(5):

    # Clear previous gradients
    optimizer.zero_grad()

    # Forward pass / Loss calculation
    loss = (w - 5) ** 2

    # Backpropagation: Calculate gradients
    loss.backward()

    # Optimizer: Update weight
    optimizer.step()

    print(
        f"Epoch: {epoch+1}, "
        f"Weight: {w.item():.4f}, "
        f"Loss: {loss.item():.4f}"
    )
````

nn.Parameter(...) "This tensor is a trainable parameter of the neural network."

## Metrics
A metric is a function that is used to judge the performance of your model.

Metric functions are similar to loss functions, except that the results from evaluating a metric are not used when training the model. Note that you may use any loss function as a metric. Loss tells us how much error the model has during training. Metrics tell us how good the model is at making predictions.

#### Q. Where are metrics used in neural network training?
Metrics are generally calculated after the forward pass, when predictions are available. Metrics can be calculated during training for monitoring, but their most important use is during validation and testing.

### 1. Forward Propagation
`Input → Model → Predictions`

↓

| Training Branch (A) | Evaluation Branch (B) |
| :--- | :--- |
| **2A. Loss Function**<br>Calculate loss | **2B. Metrics**<br>Evaluate predictions |
| ↓ | ↓ |
| **3A. Backpropagation**<br>Calculate gradients | **3B. Evaluation**<br>Accuracy, F1, etc. |
| ↓ | |
| **4A. Optimizer**<br>Update weights | |

> *The metrics branch is for monitoring and evaluation. It does not normally participate in backpropagation.*

### Which metrics should you choose?

Here is a practical selection guide:

| Your ML task | Recommended metrics |
| :--- | :--- |
| Binary classification | Accuracy, Precision, Recall, F1, ROC-AUC |
| Imbalanced classification | Precision, Recall, F1, PR-AUC |
| Multi-class classification | Accuracy, Macro F1, Weighted F1 |
| House price prediction | MAE, RMSE, $R^2$ |
| Traffic forecasting | MAE, RMSE, MAPE (when appropriate) |
| Fault detection | Recall, Precision, F1 |
| Wi-Fi AP load prediction | MAE, RMSE, $R^2$ |
| Wi-Fi client classification | Accuracy, Precision, Recall, F1 |

## Loss Functions in Neural Networks

A **Loss Function** measures the difference between the model's predicted output and the actual target value. It tells the neural network how wrong its predictions are. The loss value is used during backpropagation to calculate gradients, which the optimizer uses to update model weights.

#### Regression Losses

Used when predicting continuous numerical values.

Examples:
* House price prediction
* Temperature forecasting
* Traffic prediction
* Wi-Fi AP load prediction
![Regression Losses](image-6.png)

### Classification Losses

Used when predicting categories or classes.

Examples:
* Spam / Not Spam
* Cat / Dog
* Fraud / Not Fraud
* Fault / No Fault

![Classification Losses](image-7.png)
