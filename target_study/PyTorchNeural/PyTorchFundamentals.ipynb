NumPy arrays and PyTorch tensors both store and process numerical data. The main difference is that PyTorch tensors support GPU acceleration and automatic differentiation (Autograd), while NumPy arrays primarily operate on the CPU.

##### What is broadcasting in PyTorch?

`Broadcasting in PyTorch` is a mechanism that allows mathematical `operations between tensors of different shapes` without manually resizing or copying their data. PyTorch automatically expands dimensions logically when their shapes are compatible.

**Why is broadcasting useful?**

Avoids manually repeating tensor values. `Reduces unnecessary memory allocation`. Useful in neural networks, normalization and batch processing.

**Broadcasting compatibility:**

| Tensor A | Tensor B | Broadcasting |
| :--- | :--- | :--- |
| (2, 3) | (2, 3) | Possible |
| (2, 3) | (1, 3) | Possible |
| (2, 3) | (2, 1) | Possible |
| (2, 3) | (4, 3) | Not possible |
| (2, 3) | (2, 4) | Not possible |
| (3, 4) | (2, 3, 4) | Possible |

##### What actually happen in broadcasting ?

```python
import torch
a = torch.tensor([[1, 2, 3],
                  [4, 5, 6]])
b = torch.tensor([10, 20, 30])
c = a + b
print(c)
```
Output:
```
tensor([[11, 22, 33],
        [14, 25, 36]])
```
$\textcolor{red}{\textbf{PyTorch}}$ broadcasts $\textcolor{blue}{\textbf{b}}$ across both rows of $\textcolor{green}{\textbf{a}}$.


##### What is the difference between `view()`, `reshape()` and `flatten()`?

- `view()` Changes tensor shape without copying data. It requires the tensor to have a compatible memory layout (contiguous).
- `reshape()` Changes tensor shape. It returns a view when possible; otherwise, it creates a copy.
- `flatten()` Converts a tensor into a 1D tensor by flattening its dimensions.


##### What is the difference between `torch.cat()` and `torch.stack()`?

In PyTorch, both `torch.cat()` and `torch.stack()` combine multiple tensors
* `torch.cat()` joins tensors along an existing dimension. dimension not change before and after.
* `torch.stack()` joins tensors along a new dimension. dimension changed after operations.

````python
import torch

a = torch.tensor([[1, 2], [3, 4]])
b = torch.tensor([[5, 6], [7, 8]])
print("Concatenation with cat & dim=0")
c = torch.cat((a, b), dim=0)
print(c)
print(c.shape,c.ndim)
print("Concatenation with cat & dim=1")
c = torch.cat((a, b), dim=1)
print(c)
print(c.shape,c.ndim)

print("Concatenation with cat & dim=0")
c = torch.stack((a, b), dim=0)
print(c)
print(c.shape,c.ndim)

print("Concatenation with stack & dim=1")
c = torch.stack((a, b), dim=1)
print(c)
print(c.shape,c.ndim)

print("Concatenation with stack & dim=0")
c = torch.stack((a, b), dim=0)
print(c)
print(c.shape,c.ndim)
````

#### What is the difference between `clone()` and `detach()`?

In $\textcolor{red}{\textbf{PyTorch}}$, $\textcolor{blue}{\texttt{clone()}}$ creates a $\textcolor{green}{\textbf{copy of tensor data}}$, whereas $\textcolor{purple}{\texttt{detach()}}$ separates a tensor from the $\textcolor{orange}{\textbf{computational graph (Autograd)}}$.
* `clone()` → Copy data. Creates a new tensor with separate memory but maintains the gradient relationship with the original tensor.
* `detach()` → Stop gradient tracking. Creates a tensor sharing the original data but disconnected from its computational graph.
* `clone().detach()` → Independent copy without Autograd tracking.

Use $\textcolor{blue}{\texttt{clone()}}$ when you want to $\textcolor{green}{\textbf{modify a tensor}}$ without $\textcolor{red}{\textbf{modifying the original tensor}}$, while $\textcolor{purple}{\textbf{preserving gradient tracking}}$.

Use $\textcolor{purple}{\texttt{detach()}}$ when you want to $\textcolor{red}{\textbf{stop gradient tracking}}$.

##### **PyTorch: What happens when a tensor is detached from the computation graph?**

When a tensor is detached using `.detach()`, PyTorch disconnects that tensor from the computational graph.

1. The detached tensor no longer tracks gradients.
2. Gradients cannot flow backward through the detached tensor to previous operations.
3. The original tensor and its computational graph remain unchanged.
4. The detached tensor shares the same underlying memory with the original tensor.


````python
import torch
x = torch.tensor([1., 2., 3.], requires_grad=True)
y = x.clone()
#print(y)
print(f'Check pointer for both variable :{y.data_ptr() == x.data_ptr()}')  
print(f' gradient relationship with the original tensor : {y.requires_grad}') 

print('======= Remove gradient tracking ============')
y = x.detach()
#print(y)
print(f'Check pointer for both variable :{y.data_ptr() == x.data_ptr()}')  
print(f' gradient relationship with the original tensor : {y.requires_grad}') 

print('======= clone().detach() creates an independent copy without gradient tracking. ============')
y = x.clone().detach()
print(f'Check pointer for both variable :{y.data_ptr() == x.data_ptr()}')  
print(f' gradient relationship with the original tensor : {y.requires_grad}')  
````

##### What is the difference between contiguous and non-contiguous tensors?

contiguous and non-contiguous tensors describe how tensor elements are arranged in memory. A `contiguous` tensor stores its elements in a continuous, sequential memory layout. Use `.contiguous()` to create a contiguous copy when necessary. A `non-contiguous tensor` does `not follow `the standard sequential memory layout. This commonly happens after operations such as `transpose()` and `permute()`.

`Strides` Strides tell PyTorch `how many memory positions to move` to access the next element along each dimension.
```python
x = torch.arange(6).reshape(2, 3)
y = x.transpose(0, 1)

print(x.stride())  # (3, 1)
print(y.stride())  # (1, 3)
# For x, moving across columns requires 1 memory position; moving across rows requires 3.
```
