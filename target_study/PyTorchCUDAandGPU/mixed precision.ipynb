
## Mixed precision

Mixed precision means using multiple numerical formats during model training instead of performing every operation using FP32.

For example:

* Model computations: BF16 or FP16.
* Certain sensitive calculations: FP32.
* Optimizer states: often FP32, depending on implementation.

The objective is to reduce memory consumption and improve computational performance while maintaining numerical stability.

#### What are FP16 and BF16?

FP16 and BF16 are two 16-bit floating-point data formats used in deep learning to reduce GPU memory consumption and accelerate computations.

Both use **16 bits (2 bytes)** per value, compared with FP32, which uses 32 bits (4 bytes).

The main difference is how those 16 bits are distributed.

**FP32 vs FP16 vs BF16**

| Format | Sign bits | Exponent bits | Mantissa bits | Memory |
| :--- | ---: | ---: | ---: | ---: |
| FP32 | 1 | 8 | 23 | 4 bytes |
| FP16 | 1 | 5 | 10 | 2 bytes |
| BF16 | 1 | 8 | 7 | 2 bytes |

#### Difference between FP16 and BF16

The important difference is numerical range and precision.

| Feature | FP16 | BF16 |
| :--- | :--- | :--- |
| Memory | 2 bytes | 2 bytes |
| Numerical range | Smaller | Similar to FP32 |
| Precision | Higher than BF16 | Lower than FP16 |
| Overflow risk | Higher | Lower |
| Training stability | May require loss scaling | Generally more stable |
| LLM training | Supported | Often preferred |

**Why is BF16 often preferred for LLM training?**

BF16 has the same number of exponent bits as FP32. Therefore, it can represent a much wider range of values than FP16. This helps prevent numerical overflow and underflow during training.
FP16 has more mantissa bits, so it provides better precision for values within its smaller numerical range.
